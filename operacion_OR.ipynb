```python
# NEURONA McCulloch-Pitts: OR
def neurona_OR(a, b):
    peso = 1
    umbral = 1
    suma = a * peso + b * peso
    return 1 if suma >= umbral else 0

# Pruebas
print("0 O 0 =", neurona_OR(0,0))
print("0 O 1 =", neurona_OR(0,1))
print("1 O 0 =", neurona_OR(1,0))
print("1 O 1 =", neurona_OR(1,1))
```

Lo que hace la neurona: Se enciende (da 1) cuando al menos una de las entradas es 1. Solo se apaga cuando las dos son 0. Es como decir "sí si al menos uno es sí".
